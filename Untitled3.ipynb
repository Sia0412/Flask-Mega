```python
#@title target distribution
target = "Patient_Health_Status"

if target in df.columns:

    print("\nTarget Class Counts")

    display(df[target].value_counts())

    print("\nTarget Class Percentage")

    display(round(df[target].value_counts(normalize=True)*100,2))

else:
    print(f"{target} column not found.")
```

```python
#@title Feature Ranges
print("\nFeature Ranges")

numeric_columns = df.select_dtypes(include=np.number).columns

feature_ranges = pd.DataFrame({
    'Minimum': df[numeric_columns].min(),
    'Maximum': df[numeric_columns].max(),
    'Range': df[numeric_columns].max() - df[numeric_columns].min()
})

display(feature_ranges)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

if target in df.columns:

    plt.figure(figsize=(7,5))

    sns.countplot(
        data=df,
        x=target,
        order=df[target].value_counts().index
    )

    plt.title("Patient Health Status Distribution")
    plt.xlabel("Health Status")
    plt.ylabel("Count")
    plt.xticks(rotation=20)

    plt.show()

print("\nStep 1 Completed Successfully.")
```

```python
target = "Patient_Health_Status"

print("="*60)
print("STEP 2: TARGET VALIDATION AND ENCODING")
print("="*60)
if target not in df.columns:
    raise ValueError(f"'{target}' column was not found in the dataset.")

print(f"'{target}' column found successfully.")
```

```python
#@title class counts and percentage
print("\nClass Counts:")
class_counts = df[target].value_counts().sort_index()
print(class_counts)
```

```python
print("\nClass Percentages (%):")
class_percentages = round(df[target].value_counts(normalize=True).sort_index() * 100, 2)
print(class_percentages)
```

```python
#@title Encode target labels
label_mapping = {
    "Stable": 0,
    "Moderate_Risk": 1,
    "High_Risk": 2,
    "Critical": 3
}

# Keep original labels for later interpretation
df["Patient_Health_Status_Label"] = df[target]

# Encode target
df[target] = df[target].map(label_mapping)
```

```python
if df[target].isnull().sum() > 0:
    print("\nWARNING!")
    print("Some labels were not encoded.")
    print(df[df[target].isnull()]["Patient_Health_Status_Label"].unique())
else:
    print("\nAll target labels encoded successfully.")
```

```python
#@title Prepare feature matrix and target vector
X = df.drop(columns=[
    "Patient_Health_Status",          # encoded target
    "Patient_Health_Status_Label"     # original labels
])
y = df["Patient_Health_Status"]

print("\nFeature Matrix Shape :", X.shape)
print("Target Shape         :", y.shape)

print("\nTarget column has been excluded from input features.")

print("\nStep 2 Completed Successfully.")
```
